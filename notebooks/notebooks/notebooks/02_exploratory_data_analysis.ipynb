```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

pd.set_option("display.max_columns", None)

data_path = next(
    path for path in [
        Path("../data/raw/application_train.csv"),
        Path("../../data/raw/application_train.csv"),
    ]
    if path.exists()
)

df = pd.read_csv(data_path)

print(df.shape)
df.head()
```

```python
target_counts = df["TARGET"].value_counts()

print(target_counts)

print("\nPercentage:")
print(df["TARGET"].value_counts(normalize=True) * 100)
```

```python
plt.figure(figsize=(6, 4))

sns.countplot(data=df, x="TARGET")

plt.title("Target Distribution")
plt.xlabel("Repayment Difficulty")
plt.ylabel("Number of Customers")

plt.show()
```

```python
df["AGE_YEARS"] = (-df["DAYS_BIRTH"] / 365.25).round(1)

df[["DAYS_BIRTH", "AGE_YEARS"]].head()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="AGE_YEARS",
    bins=30,
    kde=True
)

plt.title("Customer Age Distribution")
plt.xlabel("Age in Years")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="TARGET",
    y="AGE_YEARS"
)

plt.title("Age vs Repayment Difficulty")
plt.xlabel("TARGET")
plt.ylabel("Age")

plt.show()
```

```python
df["AMT_INCOME_TOTAL"].describe()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    np.log1p(df["AMT_INCOME_TOTAL"]),
    bins=50,
    kde=True
)

plt.title("Log Distribution of Customer Income")
plt.xlabel("log(Income + 1)")

plt.show()
```

```python
income_by_target = df.groupby("TARGET")["AMT_INCOME_TOTAL"].agg(
    ["mean", "median", "min", "max"]
)

income_by_target
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="TARGET",
    y=np.log1p(df["AMT_INCOME_TOTAL"])
)

plt.title("Income vs Repayment Difficulty")
plt.xlabel("TARGET")
plt.ylabel("Log Income")

plt.show()
```

```python
loan_columns = [
    "AMT_CREDIT",
    "AMT_ANNUITY",
    "AMT_GOODS_PRICE"
]

df[loan_columns].describe()
```

```python
for column in loan_columns:
    
    plt.figure(figsize=(8, 4))
    
    sns.histplot(
        data=df,
        x=column,
        bins=40
    )
    
    plt.title(f"Distribution of {column}")
    
    plt.show()
```

```python
df.groupby("TARGET")[loan_columns].median()
```

```python
gender_risk = pd.crosstab(
    df["CODE_GENDER"],
    df["TARGET"],
    normalize="index"
) * 100

gender_risk
```

```python
plt.figure(figsize=(7, 4))

sns.barplot(
    data=df,
    x="CODE_GENDER",
    y="TARGET",
    estimator="mean"
)

plt.title("Repayment Difficulty Rate by Gender")
plt.ylabel("Default Risk Rate")

plt.show()
```

```python
education_risk = (
    df.groupby("NAME_EDUCATION_TYPE")["TARGET"]
    .agg(["count", "mean"])
    .sort_values("mean", ascending=False)
)

education_risk
```

```python
education_order = (
    df.groupby("NAME_EDUCATION_TYPE")["TARGET"]
    .mean()
    .sort_values(ascending=False)
    .index
)

plt.figure(figsize=(10, 5))

sns.barplot(
    data=df,
    x="NAME_EDUCATION_TYPE",
    y="TARGET",
    order=education_order
)

plt.xticks(rotation=30)

plt.title("Repayment Difficulty Rate by Education")

plt.show()
```

```python
df["DAYS_EMPLOYED"].describe()
```

```python
df["DAYS_EMPLOYED_CLEAN"] = df["DAYS_EMPLOYED"].replace(
    365243,
    np.nan
)
```

```python

df["EMPLOYMENT_YEARS"] = (
    -df["DAYS_EMPLOYED_CLEAN"] / 365.25
).round(1)

df["EMPLOYMENT_YEARS"].describe()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="EMPLOYMENT_YEARS",
    bins=40
)

plt.xlim(0, 50)

plt.title("Employment Duration")

plt.show()
```

```python
df.groupby("TARGET")["EMPLOYMENT_YEARS"].median()
```

```python
family_risk = (
    df.groupby("NAME_FAMILY_STATUS")["TARGET"]
    .agg(["count", "mean"])
    .sort_values("mean", ascending=False)
)

family_risk
```

```python
family_order = (
    df.groupby("NAME_FAMILY_STATUS")["TARGET"]
    .mean()
    .sort_values(ascending=False)
    .index
)

plt.figure(figsize=(10, 5))

sns.barplot(
    data=df,
    x="NAME_FAMILY_STATUS",
    y="TARGET",
    order=family_order
)

plt.xticks(rotation=30)

plt.title("Repayment Difficulty by Family Status")

plt.show()
```

```python
occupation_risk = (
    df.groupby("OCCUPATION_TYPE")["TARGET"]
    .agg(["count", "mean"])
    .sort_values("mean", ascending=False)
)

occupation_risk.head(15)
```

```python
occupation_filtered = occupation_risk[
    occupation_risk["count"] >= 500
].sort_values(
    "mean",
    ascending=False
)

occupation_filtered
```

```python
plt.figure(figsize=(10, 8))

sns.barplot(
    x=occupation_filtered["mean"],
    y=occupation_filtered.index
)

plt.title(
    "Repayment Difficulty by Occupation\n(Minimum 500 Customers)"
)

plt.xlabel("Repayment Difficulty Rate")

plt.show()
```

```python
selected_features = [
    "TARGET",
    "AMT_INCOME_TOTAL",
    "AMT_CREDIT",
    "AMT_ANNUITY",
    "AMT_GOODS_PRICE",
    "AGE_YEARS",
    "EMPLOYMENT_YEARS",
    "CNT_CHILDREN",
    "CNT_FAM_MEMBERS"
]

correlation = df[selected_features].corr()

correlation
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Between Selected Features")

plt.show()
```

Key EDA Findings

1. The target variable is highly imbalanced.

2. Age appears to have some relationship with repayment difficulty.

3. Income contains significant outliers, making the median and log-transformed
   distribution more informative than raw visualization.

4. DAYS_EMPLOYED contains the sentinel value 365243, which must be handled
   during preprocessing.

5. Some categorical features, such as education, occupation, and family status,
   show different repayment difficulty rates.

6. Categories with small sample sizes should not be interpreted without
   considering their observation count.
